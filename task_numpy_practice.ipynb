 
 
* __Задача 6__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Дополнительно сравните с функцией scipy.spatial.distance.cdist по скорости работы (сравнения приведите ниже в ноутбуке).

```python
import functions
import functions_vectorized

from importlib import reload
from scipy.spatial.distance import cdist

def reload_solution_modules():
    """Reload the two local solution modules after their files change."""
    reload(functions)
    reload(functions_vectorized)

reload_solution_modules() 

import timeit
import matplotlib.pyplot as plt
import numpy as np

sizes = [100, 300, 500, 800, 1000]
d = 3 # размерность

times_pure, times_vec, times_cdist = [], [], []

for n in sizes:
    X = np.random.randint(0, 10, size=(n, d))
    Y = np.random.randint(0, 10, size=(n, d))

    t_pure = timeit.timeit(lambda: functions.pairwise_distance(X, Y), number=3) / 3
    t_vec = timeit.timeit(lambda: functions_vectorized.pairwise_distance(X, Y), number=100) / 100
    t_cdist = timeit.timeit(lambda: cdist(X, Y, metric='euclidean'), number=100) / 100

    times_pure.append(t_pure)
    times_vec.append(t_vec)
    times_cdist.append(t_cdist)

plt.figure(figsize=(8, 5))
plt.plot(sizes, times_pure, marker='o', label='Без numpy (циклы)')
plt.plot(sizes, times_vec,   marker='s', label='Векторизованная (numpy)')
plt.plot(sizes, times_cdist, marker='^', label='scipy.cdist')
plt.xlabel('Количество объектов (n)')
plt.xticks(sizes)
plt.ticklabel_format(axis='x', style='plain', useOffset=False)
plt.ylabel('Время (с)')
plt.title('Сравнение скорости: pairwise_distance')
plt.grid(True, alpha=0.3)
plt.yscale('log')
plt.legend()
plt.tight_layout()
plt.show()
```
